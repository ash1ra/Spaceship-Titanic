```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("data/train.csv")
```

```python
cabin_splitted = df["Cabin"].str.split("/", expand=True)
df["CabinDeck"] = cabin_splitted[0]
df["CabinNum"] = cabin_splitted[1]
df["CabinSide"] = cabin_splitted[2]
```

```python
df["HomePlanet"] = df["HomePlanet"].astype("category")
df["Destination"] = df["Destination"].astype("category")
df["CabinDeck"] = df["CabinDeck"].astype("category")
df["CabinSide"] = df["CabinSide"].astype("category")
df["CabinNum"] = df["CabinNum"].astype("float")
df["CryoSleep"] = df["CryoSleep"].astype("bool")
df["VIP"] = df["VIP"].astype("bool")
```

```python
df["HomePlanetCode"] = df["HomePlanet"].cat.codes
df["DestinationCode"] = df["Destination"].cat.codes
df["CabinDeck"] = df["CabinDeck"].cat.codes
df["CabinSide"] = df["CabinSide"].cat.codes

df = df.replace({True: 1, False: 0})

df["CryoSleep"] = df["CryoSleep"].astype("int8")
df["VIP"] = df["VIP"].astype("int8")
df["Transported"] = df["Transported"].astype("int8")
```

```python
df = df[["HomePlanetCode", "DestinationCode", "Age", "CryoSleep", "VIP", "RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck", "CabinDeck", "CabinNum", "CabinSide", "Transported"]]
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.isna().sum() / df.shape[0] * 100
```

```python
num_cols = ["Age", "RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck", "CabinNum"]

df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
df["HomePlanetCode"] = df["HomePlanetCode"].fillna(df["HomePlanetCode"].value_counts().index[0])
df["DestinationCode"] = df["DestinationCode"].fillna(df["DestinationCode"].value_counts().index[0])
df["CabinDeck"] = df["CabinDeck"].fillna(df["CabinDeck"].value_counts().index[0])
df["CabinSide"] = df["CabinSide"].fillna(df["CabinSide"].value_counts().index[0])
```

```python
df.isna().sum()
```

```python
y = df["Transported"]
X = df.drop(columns=["Transported"])
```

```python
#cols_to_scale = ["RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck", "CabinNum"]

scaler = StandardScaler().fit(X)
X_scaled = pd.DataFrame(scaler.transform(X))
X_scaled.columns = X.columns
```

```python
X_scaled
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_scaled, y)
```

```python
test_df = pd.read_csv("data/test.csv")

cabin_splitted = test_df["Cabin"].str.split("/", expand=True)
test_df["CabinDeck"] = cabin_splitted[0]
test_df["CabinNum"] = cabin_splitted[1]
test_df["CabinSide"] = cabin_splitted[2]

test_df["HomePlanet"] = test_df["HomePlanet"].astype("category")
test_df["Destination"] = test_df["Destination"].astype("category")
test_df["CabinDeck"] = test_df["CabinDeck"].astype("category")
test_df["CabinSide"] = test_df["CabinSide"].astype("category")
test_df["CabinNum"] = test_df["CabinNum"].astype("float")
test_df["CryoSleep"] = test_df["CryoSleep"].astype("bool")
test_df["VIP"] = test_df["VIP"].astype("bool")

test_df["HomePlanetCode"] = test_df["HomePlanet"].cat.codes
test_df["DestinationCode"] = test_df["Destination"].cat.codes
test_df["CabinDeck"] = test_df["CabinDeck"].cat.codes
test_df["CabinSide"] = test_df["CabinSide"].cat.codes

test_df = test_df.replace({True: 1, False: 0})

test_df["CryoSleep"] = test_df["CryoSleep"].astype("int8")
test_df["VIP"] = test_df["VIP"].astype("int8")

test_df_pid = test_df["PassengerId"]

test_df = test_df[["HomePlanetCode", "DestinationCode", "Age", "CryoSleep", "VIP", "RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck", "CabinDeck", "CabinNum", "CabinSide"]]

num_cols = ["Age", "RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck", "CabinNum"]

test_df[num_cols] = test_df[num_cols].fillna(test_df[num_cols].mean())

test_df["HomePlanetCode"] = test_df["HomePlanetCode"].fillna(test_df["HomePlanetCode"].value_counts().index[0])
test_df["DestinationCode"] = test_df["DestinationCode"].fillna(test_df["DestinationCode"].value_counts().index[0])
test_df["CabinDeck"] = test_df["CabinDeck"].fillna(test_df["CabinDeck"].value_counts().index[0])
test_df["CabinSide"] = test_df["CabinSide"].fillna(test_df["CabinSide"].value_counts().index[0])

test_df_scaled = pd.DataFrame(scaler.transform(test_df))
test_df_scaled.columns = test_df.columns
```

```python
predictions = model.predict(test_df_scaled)
```

```python
submission_df = pd.DataFrame({"PassengerId": test_df_pid, "Transported": pd.Series(predictions).astype(bool)})
submission_df
```

```python
submission_df.to_csv("submission.csv", index=False)
```
